## Phase 2 — Notebook initialization and workspace setup

````python
from pathlib import Path
from config import LAST_NAME, STUDENT_ID
from lab6_core import (
    make_student, create_workspace, task7_8_notes, task9_10_read_methods,
    task11_filter_notes, task12_csv, task13_json, task14_metadata_backup,
    task15_preview_dataset, task16_process_sample, task17_write_report,
    render_observations,
)

# Select DEMO mode explicitly for a code test; keep False for submitted results.
USE_DEMO_DATA = False
WORK_DIR = Path.cwd().resolve()
if not (WORK_DIR / 'config.py').exists():
    raise RuntimeError('Open this notebook with the repository root as the working directory.')
student = make_student(LAST_NAME, STUDENT_ID)
LAST_NAME, STUDENT_ID = student.last_name, student.student_id
SEED_DIGIT, ID_SUM, NAME_LENGTH = student.seed_digit, student.id_sum, student.name_length
verification_value = student.verification_value
print('Student:', LAST_NAME)
print('Student ID:', STUDENT_ID)
print('Seed:', SEED_DIGIT, '| ID sum:', ID_SUM, '| Name length:', NAME_LENGTH)
print('Verification value:', verification_value)
````

### Task 6 — Create and check the data, output, and backup folders

````python
dirs = create_workspace(WORK_DIR)
DATA_DIR, OUTPUT_DIR, BACKUP_DIR = dirs['data'], dirs['output'], dirs['backup']
print('Workspace:', WORK_DIR)
for name, folder in dirs.items():
    print(f'{name} folder exists: {folder.exists()} — {folder}')
````

## Phase 3 — Basic file I/O

### Task 7 — Examine paths and create a text file

````python
file_path = task7_8_notes(WORK_DIR, student)  # also performs Task 8 (append)
print('Full path:', file_path)
print('File name:', file_path.name)
print('File extension:', file_path.suffix)
print('Parent folder:', file_path.parent)
print('Created:', file_path.exists())
````

### Task 8 — Append verification (without erasing the original text)

````python
print('Verification value:', verification_value)
with open(file_path, 'r', encoding='utf-8') as file:
    print(file.read())
print('The helper used "w" to create the original and "a" to append the verification line.')
````

### Task 9 — Compare `read()`, `readline()`, and `readlines()`

````python
methods = task9_10_read_methods(file_path)
print('read():', repr(methods['read']))
print('readline():', repr(methods['readline']))
print('readlines():', repr(methods['readlines']))
````

### Task 10 — Read the file one line at a time

````python
for line_number, text in methods['line_by_line']:
    print(f'Line {line_number}: {text}')
````

## Phase 4 — Filtering and structured data

### Task 11 — Filter lines containing `ID` or `Verification`

````python
filtered_path, filtered_count = task11_filter_notes(file_path, OUTPUT_DIR, student)
print('Filtered file:', filtered_path)
print('Selected lines:', filtered_count)
print(filtered_path.read_text(encoding='utf-8'))
````

### Task 12 — Create the student-specific CSV record

````python
csv_path = task12_csv(OUTPUT_DIR, student)
print('Created CSV:', csv_path)
print('Single data row (name, ID, seed, ID sum):')
print(csv_path.read_text(encoding='utf-8'))
````

### Task 13 — Write structured JSON and verify the stored value

````python
json_path, expected_verification, json_status = task13_json(OUTPUT_DIR, student)
print('Created JSON:', json_path)
print(json_path.read_text(encoding='utf-8'))
print('Stored/expected verification:', verification_value, '/', expected_verification)
print('Status:', json_status)
````

## Phase 5 — File management

### Task 14 — Inspect CSV metadata and verify a backup

````python
metadata = task14_metadata_backup(csv_path, BACKUP_DIR)
for key, value in metadata.items():
    print(f'{key}: {value}')
````

## Phase 6 — Real-world application and report

### Task 15 — Preview the UCI household power file

````python
dataset_path = (
    WORK_DIR / 'sample_data' / 'household_power_consumption_DEMO.txt'
    if USE_DEMO_DATA else DATA_DIR / 'household_power_consumption.txt'
)
if USE_DEMO_DATA:
    print('*** DEMONSTRATION ONLY: values are NOT from the original UCI dataset. ***')
header, sample_records = task15_preview_dataset(dataset_path)
print('Dataset:', dataset_path)
print('Header fields:', header)
for i, sample in enumerate(sample_records, 1):
    print(f'Sample record {i}: {sample}')
````

### Task 16 — Process a `50 + ID_SUM` record sample and calculate mean Global Active Power

````python
processing = task16_process_sample(dataset_path, student)
print('Sample size:', processing['sample_size'])
print('Rows examined:', processing['records_examined'])
print('Valid records:', processing['valid_records'])
print('Invalid / missing power records:', processing['invalid_records'])
print('Average Global Active Power:', round(processing['average_power'], 3))
if processing['valid_records'] == 0:
    print('WARNING: No numeric Global Active Power measurements in this sample.')
````

### Task 17 — Generate the persistent text report

````python
report_path = task17_write_report(OUTPUT_DIR, student, processing, dataset_path)
print('Report created:', report_path)
print(report_path.read_text(encoding='utf-8'))
````
