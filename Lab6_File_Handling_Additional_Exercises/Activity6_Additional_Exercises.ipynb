# Activity 6 — Additional Exercises (Pages 14–16)

The three extra exercises use Python's **standard library**. Download the original Kaggle datasets into `data/` (paths in the root README), and edit `config.py` before running. The scripts **stream** web logs and validate mobile/cyber CSV values. Optional synthetic fixtures are available only for code tests.

**Deployment:** create/push a separate Git branch `activity6-additional-exercises` as directed in the handout.

```python
from pathlib import Path
import sys
ROOT = Path.cwd().resolve()
if not (ROOT / 'config.py').exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))
from config import LAST_NAME, STUDENT_ID
from lab6_core import make_student
from Lab6_File_Handling_Additional_Exercises.additional_exercises import (
    web_server_logs, mobile_device_usage, cybersecurity_incidents,
)
student = make_student(LAST_NAME, STUDENT_ID)
# Set to True only for a test with tiny synthetic fixtures.
USE_DEMO_DATA = False
source_dir = ROOT / ('sample_data' if USE_DEMO_DATA else 'data')
out_dir = ROOT / 'output' / 'additional_exercises'
backup_dir = ROOT / 'backup' / 'additional_exercises'
print('Student:', student)
if USE_DEMO_DATA:
    print('*** DEMO DATA ONLY — not genuine Kaggle statistics ***')
```

## Exercise 1 — Web Server Access Log Analysis

```python
web_path = source_dir / ('weblog_DEMO.csv' if USE_DEMO_DATA else 'weblog.csv')
web = web_server_logs(web_path, out_dir, backup_dir, student)
print('Records processed:', web['records_processed'])
print('Target status class:', web['target_status_class'])
print('Records in target class:', web['records_in_target_class'])
print('Count per HTTP status code:', web['count_per_status_code'])
print('Top 5 requested URLs:', web['five_most_requested_urls'])
print('Filtered CSV:', web['filtered_csv'])
print('JSON summary:', web['json_summary_file'])
print('Backup verification:', web['backup_verification'])
```

## Exercise 2 — Mobile Device Usage Profiling

```python
mobile_path = source_dir / ('user_behavior_dataset_DEMO.csv' if USE_DEMO_DATA else 'user_behavior_dataset.csv')
mobile = mobile_device_usage(mobile_path, out_dir, backup_dir, student)
print('Target:', mobile['operating_system'], 'behavior class', mobile['user_behavior_class'])
print('Total records:', mobile['total_records_in_dataset'])
print('OS matches:', mobile['records_matching_operating_system'])
print('Selected records:', mobile['records_in_target_behavior_class'])
print('Average battery drain:', mobile['average_battery_drain_mAh_per_day'])
print('Average data usage:', mobile['average_data_usage_MB_per_day'])
print('Count verification:', mobile['count_verification'])
print('Files:', mobile['exported_csv_file'], mobile['json_summary_file'])
print('Backup:', mobile['backup_file_and_size'])
```

## Exercise 3 — Cybersecurity Incident Report

```python
cyber_path = source_dir / ('Global_Cybersecurity_Threats_DEMO.csv' if USE_DEMO_DATA else 'Global_Cybersecurity_Threats_2015-2024.csv')
cyber = cybersecurity_incidents(cyber_path, out_dir, backup_dir, student)
print('Target year:', cyber['target_year'])
print('Records in dataset:', cyber['total_records_in_dataset'])
print('Incidents in target year:', cyber['incidents_in_target_year'])
print('Counts by attack:', cyber['count_per_attack_type'])
print('Total loss (USD million):', cyber['total_financial_loss_usd_millions'])
print('Average loss (USD million):', cyber['average_financial_loss_usd_millions'])
print('Top five incidents:', cyber['five_costliest_incidents'])
print('Report:', cyber['report_file'])
print('Backup verified:', cyber['backup_verification']['identical'])
```
